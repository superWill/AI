# Apple cash accumulation analysis

## TL;DR

Apple did not continuously accumulate cash from FY2018 through FY2025. Gross cash and net cash fell materially. FY2026 H1 is a genuine balance-sheet reversal, but its unusually strong operating cash flow includes tax and working-capital timing effects.

## Context & methods

- Gross cash = cash and cash equivalents + current marketable securities + non-current marketable securities.
- Debt = commercial paper + current term debt + non-current term debt.
- Net cash = gross cash - debt.
- Free cash flow = operating cash flow - payments for property, plant and equipment.
- FY2026 H1 is a six-month flow period and a March 28, 2026 balance-sheet point; it is not annualized.

```python
from pprint import pprint

balance = [
    {'period': 'FY2018', 'gross_cash': 237.100, 'debt': 114.483},
    {'period': 'FY2019', 'gross_cash': 205.898, 'debt': 108.047},
    {'period': 'FY2020', 'gross_cash': 191.830, 'debt': 112.436},
    {'period': 'FY2021', 'gross_cash': 190.516, 'debt': 124.719},
    {'period': 'FY2022', 'gross_cash': 169.109, 'debt': 120.069},
    {'period': 'FY2023', 'gross_cash': 162.099, 'debt': 111.088},
    {'period': 'FY2024', 'gross_cash': 156.650, 'debt': 106.629},
    {'period': 'FY2025', 'gross_cash': 132.420, 'debt': 98.657},
    {'period': 'FY2026 H1', 'gross_cash': 146.595, 'debt': 84.711},
]
for row in balance:
    row['net_cash'] = round(row['gross_cash'] - row['debt'], 3)
pprint(balance)
```

```python
flows = [
    {'period': 'FY2023', 'period_type': 'FY', 'ocf': 110.543, 'capex': 10.959, 'buybacks': 77.550, 'dividends': 15.025},
    {'period': 'FY2024', 'period_type': 'FY', 'ocf': 118.254, 'capex': 9.447, 'buybacks': 94.949, 'dividends': 15.234},
    {'period': 'FY2025', 'period_type': 'FY', 'ocf': 111.482, 'capex': 12.715, 'buybacks': 90.711, 'dividends': 15.421},
    {'period': 'FY2025 H1', 'period_type': 'H1', 'ocf': 53.887, 'capex': 6.011, 'buybacks': 49.504, 'dividends': 7.614},
    {'period': 'FY2026 H1', 'period_type': 'H1', 'ocf': 82.627, 'capex': 4.344, 'buybacks': 36.989, 'dividends': 7.743},
]
for row in flows:
    row['fcf'] = round(row['ocf'] - row['capex'], 3)
    row['fcf_after_buybacks_dividends'] = round(row['fcf'] - row['buybacks'] - row['dividends'], 3)
pprint(flows)
```

## Results

```python
by_period = {row['period']: row for row in balance}
flows_by_period = {row['period']: row for row in flows}
fy18 = by_period['FY2018']
fy25 = by_period['FY2025']
h126 = by_period['FY2026 H1']
h125_flow = flows_by_period['FY2025 H1']
h126_flow = flows_by_period['FY2026 H1']

metrics = {
    'FY18_to_FY25_gross_cash_change_pct': (fy25['gross_cash'] / fy18['gross_cash'] - 1) * 100,
    'FY18_to_FY25_net_cash_change_pct': (fy25['net_cash'] / fy18['net_cash'] - 1) * 100,
    'FY25_to_FY26H1_gross_cash_change_pct': (h126['gross_cash'] / fy25['gross_cash'] - 1) * 100,
    'FY25_to_FY26H1_debt_change_pct': (h126['debt'] / fy25['debt'] - 1) * 100,
    'FY25_to_FY26H1_net_cash_change_pct': (h126['net_cash'] / fy25['net_cash'] - 1) * 100,
    'H1_FCF_change_pct': (h126_flow['fcf'] / h125_flow['fcf'] - 1) * 100,
    'H1_buyback_change_pct': (h126_flow['buybacks'] / h125_flow['buybacks'] - 1) * 100,
}
pprint({key: round(value, 1) for key, value in metrics.items()})
```

```python
assert round(metrics['FY18_to_FY25_gross_cash_change_pct'], 1) == -44.2
assert round(metrics['FY18_to_FY25_net_cash_change_pct'], 1) == -72.5
assert round(metrics['FY25_to_FY26H1_gross_cash_change_pct'], 1) == 10.7
assert round(metrics['FY25_to_FY26H1_debt_change_pct'], 1) == -14.1
assert round(metrics['FY25_to_FY26H1_net_cash_change_pct'], 1) == 83.3
assert round(metrics['H1_FCF_change_pct'], 1) == 63.5
assert round(metrics['H1_buyback_change_pct'], 1) == -25.3
print('All checks passed.')
```

```python
print('period,gross_cash,debt,net_cash')
for row in balance:
    print(f"{row['period']},{row['gross_cash']:.3f},{row['debt']:.3f},{row['net_cash']:.3f}")
```

```python
print('period,fcf,buybacks,dividends')
for row in flows:
    if row['period_type'] == 'H1':
        print(f"{row['period']},{row['fcf']:.3f},{row['buybacks']:.3f},{row['dividends']:.3f}")
```

## Takeaways

1. FY2018-FY2025 was a net-cash drawdown regime, not a cash accumulation regime.
2. FY2026 H1 shows a real reversal: gross cash rose, debt fell, and buybacks slowed while free cash flow increased.
3. The balance-sheet change is more reliable than extrapolating the H1 operating-cash-flow growth rate because cash-tax and working-capital timing contributed to the year-over-year jump.
4. A policy shift away from a formal net-cash-neutral target supports the interpretation that Apple now values balance-sheet flexibility more highly, but it does not prove a specific acquisition, downturn, or AI-capex plan.
