## tl;dr

## Context & Methods

[FRAME] Scores are ordinal 0–10 judgments, not probabilities or market-share forecasts.

### Key Assumptions

[FRAME] Platform potential emphasizes developer interfaces and distribution; compute-standard potential emphasizes architecture control, deployment scale, and merchant accessibility.


## Data

[KNOWN] Primary-source links and scoring rationale are recorded in `source-notes.md`.


```python
candidates = [
    {"company": "NVIDIA", "platform_score": 9.4, "silicon_score": 9.8},
    {"company": "OpenAI", "platform_score": 9.0, "silicon_score": 3.5},
    {"company": "Microsoft", "platform_score": 8.7, "silicon_score": 4.5},
    {"company": "Alphabet", "platform_score": 8.3, "silicon_score": 6.5},
    {"company": "Amazon", "platform_score": 7.8, "silicon_score": 6.8},
    {"company": "AMD", "platform_score": 5.5, "silicon_score": 7.8},
    {"company": "Meta", "platform_score": 6.8, "silicon_score": 4.2},
    {"company": "Broadcom", "platform_score": 3.5, "silicon_score": 7.2},
]
assert len(candidates) == 8
assert all(0 <= row[key] <= 10 for row in candidates for key in ("platform_score", "silicon_score"))
candidates
```

## Results


```python
platform_rank = sorted(candidates, key=lambda row: row["platform_score"], reverse=True)
silicon_rank = sorted(candidates, key=lambda row: row["silicon_score"], reverse=True)
{
    "platform_top3": [(row["company"], row["platform_score"]) for row in platform_rank[:3]],
    "silicon_top3": [(row["company"], row["silicon_score"]) for row in silicon_rank[:3]],
}
```

## Takeaways

[INFERRED] OpenAI and Microsoft are the strongest non-NVIDIA candidates for the software/platform role, while AMD is the clearest merchant-silicon challenger.

[INFERRED] Alphabet, Amazon, and Broadcom may collectively weaken NVIDIA's economics without any one of them becoming a universal replacement.
